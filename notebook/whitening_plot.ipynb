# Draw Whitening Accuracy Plot

## 2 Layer DNN

```python
from pathlib import Path
from copy import deepcopy
import numpy as np
import sys

sys.path.append(str(Path("..")))

from src.utils import load_json
```

```python
data_list = ['mnist', 'svhn', 'cifar10', 'cifar100']
lr = '0.01'
```

```python
target = 'accuracy'
summary = dict()

for data in data_list:
    summary[data] = dict()
    summary[data]['orig'] = dict()
    summary[data]['whiten'] = dict()
    sgd_metric = list()
    sgd_result = load_json(Path("..") / 'result' / 'whitening' / f'{data}_orig.json')
    for trial in range(5):
        sgd_metric.append(sgd_result[lr][str(trial)][target])
    sgd_metric = np.stack(sgd_metric , axis=1)
    summary[data]['orig'][lr] = {
        'mean': np.mean(sgd_metric, axis=1), 
        'std': np.std(sgd_metric, axis=1)
        }
    
    vc_metric = list()
    vc_result = load_json(Path("..") / 'result' / 'whitening' / f'{data}_whiten.json')
    for trial in range(5):
        vc_metric.append(vc_result[lr][str(trial)][target])
    vc_metric = np.stack(vc_metric , axis=1)
    summary[data]['whiten'][lr] = {
        'mean': np.mean(vc_metric, axis=1),
        'std': np.std(vc_metric, axis=1)
        }
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 4, figsize=(15, 3))

for n, data in enumerate(data_list):
    orig_data = summary[data]['orig'][lr]
    whiten_data = summary[data]['whiten'][lr]
    axes[0].set_ylabel('Accuracy', fontsize=16, fontweight='bold')
    if n == 0:
        axes[n].plot(range(1, 21), orig_data['mean'], linestyle='--', color='b', label='Original', linewidth=4)
        axes[n].plot(range(1, 21), whiten_data['mean'], linestyle='-', color='r', label='Whiten', linewidth=4)
    else:
        axes[n].plot(range(1, 21), orig_data['mean'], linestyle='--', color='b', linewidth=4)
        axes[n].plot(range(1, 21), whiten_data['mean'], linestyle='-', color='r', linewidth=4)

    axes[n].fill_between(range(1, 21), orig_data['mean'] - 2 * orig_data['std'], orig_data['mean'] + 2 * orig_data['std'], color='b', alpha=0.5)
    axes[n].fill_between(range(1, 21), whiten_data['mean'] - 2 * whiten_data['std'], whiten_data['mean'] + 2 * whiten_data['std'], color='r', alpha=0.5)
    axes[n].set_xlabel(data.upper(), fontsize=14, fontweight='bold')
    axes[n].grid()

fig.legend(loc=(0.88, 0.50))
fig.supxlabel('Epochs', x=0.5, y=-0.12, fontsize=16, fontweight='bold')
fig.suptitle('Gradient Whitening', fontsize=20, fontweight='bold')
```
